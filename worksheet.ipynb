```python
def is_pelindrome(s):
    left=0
    right=len(s)-1
    while left < right:
        if s[left]!=s[right]:
            return ("false")
        left+=1
        right-=1
    return("true")
    
    
```

```python
is_pelindrome("mom")
```

```python
is_pelindrome("racecar")
```

```python
is_pelindrome("hello")
```

```python
def count_vowels(s):
    count=0
    vowels="aeiouAEIOU"
    for char in s:
        if char in vowels:
            count+=1
    return count
```

```python
count_vowels("hello world")
```

```python
count_vowels("RAJASHRI")
```

```python
def factorial(n):
    product=1
    for i in range(2,n+1):
       product*=i
    return product
```

```python
factorial(5)
```

```python
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,n):
         if n%i==0:
             return False
    return True
```

```python
is_prime(7)
```

```python
is_prime(1)
```

```python
def fizzbuzz(n):
    if n%3==0 and n%5==0:
        return ("fizzBuzz")
    elif n%3==0:
        return ("fizz")
    elif n%5==0:
        return ("Bizz")
    else:
        return str(n)
```
